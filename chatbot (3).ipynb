1. Justificación del System Prompt
He diseñado a "Gabo, el anfitrión de la Llajta". La intención es crear un chatbot con una personalidad cálida y experta, que no solo dé datos técnicos, sino que use modismos locales (como "caserita" o "pues") para una experiencia más auténtica.
Se establecieron restricciones claras para que el bot no hable de otras ciudades a menos que sea para comparar el clima o la comida, asegurando que el foco siempre sea Cochabamba. El formato de respuesta incluye siempre una recomendación gastronómica opcional, que es el corazón de la cultura cochabambina.

```python

import requests
from openai import OpenAI

HF_TOKEN = "MI_TOQUEN"
# Inicializar el cliente de OpenAI para el Hugging Face Router
client = OpenAI(
    base_url="https://router.huggingface.co/v1",
    api_key=HF_TOKEN,
)

def llamar_modelo(prompt, temperatura=0.7, model="google/gemma-7b-it"):
    # Función auxiliar para conectar con los Milestones
    try:
        completion = client.chat.completions.create(
            model=model,
            messages=[
                {
                    "role": "user",
                    "content": prompt
                }
            ],
            temperature=temperatura,
            max_tokens=500,
        )
        return completion.choices[0].message.content.strip()
    except Exception as e:
        print(f"Error de la API: {e}")
        return f"Error de conexión: {str(e)}"
```

```python
# ── MILESTONE 1: Define la personalidad de tu chatbot ────
SYSTEM_PROMPT = """
Eres Gabo, un guía turístico experto en Cochabamba, Bolivia.
Respondes siempre en español, de forma amigable, entusiasta y muy hospitalaria.

RESTRICCIONES:
1. Solo respondes sobre Cochabamba (lugares, comida, clima, historia).
2. Si te preguntan por otros países o temas fuera de Bolivia, responde amablemente que como buen cochala, solo conoces el paraíso del sol y la buena comida.
3. No des información peligrosa sobre rutas no marcadas.

FORMATO:
- Usa emojis relacionados con comida o viajes.
- Siempre termina con un "¡Se te espera en la Llajta!" o alguna frase típica similar.
"""

print("System prompt definido con éxito.")
```

```python
# ── MILESTONE 2 y 3: Chatbot con personalidad y memoria ────
historial = []

def chatbot_con_memoria(pregunta_usuario, temperatura=0.7):
    historial.append({"role": "user", "content": pregunta_usuario})

    # Construcción del prompt siguiendo el formato Zephyr solicitado
    prompt = f"<|system|>\n{SYSTEM_PROMPT}\n"
    for msg in historial:
        rol = "user" if msg["role"] == "user" else "assistant"
        prompt += f"<|{rol}|>\n{msg['content']}\n"
    prompt += "<|assistant|>"

    respuesta = llamar_modelo(prompt, temperatura)
    historial.append({"role": "assistant", "content": respuesta})
    return respuesta

def reiniciar_chat():
    global historial
    historial = []
    print("✓ Conversación reiniciada")
```

```python
# ── MILESTONE 4 y 5: Loop interactivo y Estadísticas ────

def stats_conversacion():
    turnos = len([m for m in historial if m["role"]=="user"])
    palabras = sum(len(m["content"].split()) for m in historial)
    print(f"\n📊 [STATS] Turnos: {turnos} | Palabras totales: {palabras}")

reiniciar_chat()
print("═══════════════════════════════════════")
print(" Gabo el Guía Cochala - Activo")
print("═══════════════════════════════════════\n")

# Para el notebook de entrega, simula al menos 5 turnos:
preguntas_ejemplo = [
    "¡Hola! ¿Quién eres?",
    "¿Qué puedo comer un domingo por la mañana?",
    "¿Cuál fue el primer plato que me recomendaste?",
    "¿Hace calor allá ahora?",
    "¿Me recomiendas ir al Cristo de la Concordia?"
]

for p in preguntas_ejemplo:
    print(f"👤 Tú: {p}")
    res = chatbot_con_memoria(p)
    print(f"🤖 Gabo: {res}")
    stats_conversacion() # Milestone 5 Bonus
    print("-" * 20)
```
